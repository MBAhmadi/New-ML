```python
x = np.random.normal(5.0, 10.0, 1000)
y = np.random.normal(10.0, 20.0, 1000)
```

```python
plt.scatter(x,y, alpha=0.3);
```

```python
slope, intercept, r, p, std_err = stats.linregress(x, y)
```

```python
def myfunc(x):
  return slope * x + intercept

mymodel = list(map(myfunc, x))

plt.scatter(x, y)
plt.plot(x, mymodel)
plt.show()
```

```python
print(r)
print(p)
print(std_err)
```
