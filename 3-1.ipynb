```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.datasets import make_circles
```

```python
X, Y = make_circles(n_samples=500, noise=0.1)
X[Y==0, :] *=1.5
```

```python
print(X.shape, Y.shape)
```

```python
plt.scatter(X[Y==0, 0], X[Y==0, 1], c='green')
plt.scatter(X[Y==1, 0], X[Y==1, 1], c='blue');
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import Pipeline
```

```python
lgr = LogisticRegression()
lgr.fit(X, Y)
```

```python
steps = [('poly_feature', PolynomialFeatures(degree=2)),
         ('lgr', LogisticRegression(fit_intercept=False))]
poly_lgr = Pipeline(steps=steps)
poly_lgr.fit(X, Y)
```

```python
x1_min, x1_max = X[:, 0].min(), X[:, 0].max()
x2_min, x2_max = X[:, 1].min(), X[:, 1].max()
```

```python
xx, yy = np.meshgrid(np.linspace(x1_min, x1_max, 500),
            np.linspace(x2_min, x2_max, 500))
```

```python
xx.shape
```

```python
x_input = np.c_[xx.reshape(-1), yy.reshape(-1)]
```

```python
plt.figure(figsize=(9, 4))

plt.subplot(121)
plt.title('Logestic Regression')
plt.contour(xx, yy , lgr.predict(X=x_input).reshape(500, 500), colors='k')
y_pred = lgr.predict(X=X)
plt.scatter(X[y_pred==0, 0], X[y_pred==0, 1], c='green')
plt.scatter(X[y_pred==1, 0], X[y_pred==1, 1], c='blue');

plt.subplot(122)
plt.title('Polynomial Logestic Regression')
plt.contour(xx, yy , poly_lgr.predict(X=x_input).reshape(500, 500), colors='k')
y_pred = poly_lgr.predict(X=X)
plt.scatter(X[y_pred==0, 0], X[y_pred==0, 1], c='green')
plt.scatter(X[y_pred==1, 0], X[y_pred==1, 1], c='blue');
```
